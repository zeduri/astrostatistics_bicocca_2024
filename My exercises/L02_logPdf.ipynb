## Pdf of the log:

- Use numpy to draw N (where N is large...) random samples from a uniform distribution between 0.1 and 10, and store these samples as x.
- Use matplotlib to make a histogram of these samples.
- Compute the base-10 log of your array x, and store this as y.
- Make another histogram for y. Using the equation to transform probability distributions, write what the theoretical pdf of y is, and overplot it onto your histogram.
- Compute the log of the mean of x and the mean of y. Now compute the log of the median of x and the median of y.


You should note that the means are different, but the medians (as it is a cumulative statistic) are the same. The mean is affected by the scale of the sample values, but the median only depends on the ordering of the samples. Monotonic transformations (like taking the log) do not change the ordering of samples.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Sampling the uniform distribution:
N = int(1e5)
x = np.random.uniform(0.1, 10, N)
```

```python
#Visualizing the data:
plt.hist(x, bins = 100, histtype='step', density = True)
plt.xlabel('x')
```

```python
#Defining the new variable and making a hist: 
y = np.log10(x)
plt.hist(y, bins = 100, histtype='step', density = True)
plt.xlabel('y')
```

Given the rule to transform the pdf, we can write that: 

$\text{pdf(y)} = |dx/dy|\:\text{pdf(x)}$.

Having $y = log_{10}\:x$ and $x = 10^y$, we have $dx/dy = ln(10)\:10^y$. 

```python
samples = np.linspace(-1, 1.5, 100)
pdf = 10**samples * np.log(10)/(10-0.1)

plt.hist(y, bins = 100, histtype='step', density = True, label = 'data')
plt.plot(samples, pdf, ls = '--', label = 'Theoretical pdf')
plt.legend()
plt.xlabel('y')
plt.ylabel('P(y)')
plt.xlim(-1,1.1)
plt.ylim(0,2.5)
```

### Let's now compute the mean and the median:

```python
mean_x = np.log10(np.mean(x))
mean_y = np.mean(y)

median_x = np.log10(np.median(x))
median_y = np.median(y)

print('The log of the mean of x is:', mean_x)
print('The mean of y is:', mean_y)
print('The log of the median of x is:', median_x)
print('The median of y is:', median_y)
```

As expected, the means are different, while the median is the same, as it is a cumulative statistic.
